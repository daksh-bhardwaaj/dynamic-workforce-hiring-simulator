```python
import pandas as pd

file_path = r"C:\Users\Daksh Bhardwaj\anaconda3\Cleaned_Workforce_Dataset.xlsx"

df = pd.read_excel(file_path)

print("Shape:", df.shape)
print(df.head())

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Clean text columns
text_cols = ["Department", "Location", "Status"]

for col in text_cols:
    df[col] = df[col].str.strip().str.title()

# Check the cleaned categories
print("Departments:")
print(df["Department"].unique())

print("\nLocations:")
print(df["Location"].unique())

print("\nStatuses:")
print(df["Status"].unique())
```

```python
missing = df.isnull().sum()

print("Missing values:")
print(missing)

print("\nMissing percentage:")
print((missing / len(df) * 100).round(2))
```

```python
print("Rows with missing values:")
display(df[df.isnull().any(axis=1)].head(10))
```

```python
# Fill missing categorical values
df["Department"] = df["Department"].fillna("Unknown")
df["Location"] = df["Location"].fillna("Unknown")
df["Status"] = df["Status"].fillna("Unknown")

# Fill missing performance with median
df["Performance_Rating"] = df["Performance_Rating"].fillna(
    df["Performance_Rating"].median()
)

# Check remaining missing values
print(df.isnull().sum())
```

```python
print("Salary:")
print(df["Base_Salary_USD"].describe())

print("\nWeekly Capacity:")
print(df["Weekly_Capacity_Hrs"].describe())

print("\nPerformance Rating:")
print(df["Performance_Rating"].describe())
```

```python
print("Negative salaries:")
display(df[df["Base_Salary_USD"] < 0])

print("Unusual capacity:")
display(df[df["Weekly_Capacity_Hrs"] > 80])
```

```python
# Mark invalid salary values as missing
df.loc[df["Base_Salary_USD"] < 0, "Base_Salary_USD"] = None

# Mark invalid capacity values as missing
df.loc[df["Weekly_Capacity_Hrs"] < 0, "Weekly_Capacity_Hrs"] = None
df.loc[df["Weekly_Capacity_Hrs"] > 80, "Weekly_Capacity_Hrs"] = None

# Fill with median values
df["Base_Salary_USD"] = df["Base_Salary_USD"].fillna(
    df["Base_Salary_USD"].median()
)

df["Weekly_Capacity_Hrs"] = df["Weekly_Capacity_Hrs"].fillna(
    df["Weekly_Capacity_Hrs"].median()
)

print(df[["Base_Salary_USD", "Weekly_Capacity_Hrs"]].describe())
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nNegative salaries:")
print((df["Base_Salary_USD"] < 0).sum())

print("\nInvalid capacity:")
print(((df["Weekly_Capacity_Hrs"] < 0) |
       (df["Weekly_Capacity_Hrs"] > 80)).sum())

print("\nFinal shape:", df.shape)
```

```python
# Convert Hire_Date to datetime
df["Hire_Date"] = pd.to_datetime(
    df["Hire_Date"], errors="coerce"
)

# Check the result
print(df["Hire_Date"].dtype)

print("\nMissing Hire Dates:")
print(df["Hire_Date"].isnull().sum())

print("\nSample dates:")
print(df["Hire_Date"].head())
```

```python
print("Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nUnique Status:")
print(df["Status"].unique())

print("\nUnique Departments:")
print(df["Department"].unique())
```
